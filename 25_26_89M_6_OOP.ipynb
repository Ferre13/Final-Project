# Object oriented programming

Once we have identified the important concepts and we have create a class diagram (UML), describing the attributes (data to store) and the methods (things they can do) of every class, we can start programming them.

We will start a new file with this:
```
class NameOfTheClass:
    """ Description of the class, explaining what it is doing"""
    Code of the class
```

Rules: name of the class, we use upper camel case

A class in each file. The name of the file should be the name of the class starting by lower case: `nameOfTheClass.py`

The string in """ """ is called *docstring* and it is optional, but compulsory this year.

````python
# Our initial class, nothing is done yet
class Date:
    """ This class represents a Date to be used in my programs. It is our first
    example of OOP"""
````

# Methods

All the code inside a class must be inside a method, no code can be alone

```
def name_of_the_method(self, param2, param3, ...):
    """ Method documentation """
    instruction1
    instruction2
```

So, our classes will always be like:

```
class NameOfTheClass:
    """ Description of the class, explaining what it is doing"""

    def method1(self, param2, param3, ...):
        """ Method documentation """
        instruction1
        instruction2Code of the class

    def method2(self, param2, param3, ...):
        """ Method documentation """
        instruction1
        instruction2Code of the class

    def method3(self, param2, param3, ...):
        """ Method documentation """
        instruction1
        instruction2
```

Parts of a method:

1. Header: contains the name and the parameters (the information needed to run the method). There will be always 1 parameter, the first one, which is called `self`. The remaining ones are just variable names.
2. Documentation: explains what the method is doing
3. Body of the method: the real code.


````python
# A class to represent Mario and Luigi for the final project,
# currently it only contains the attributes
class Player:
    """ This class represents Mario or Luigi """
    def __init__(self, image : str, side: str):
        """ This is the init method of my class. Here is where the
        attributes of the objects of this class are defined """
        # Those are the attributes of the class, they will always
        # start with self.
        # For each attribute I need to think if it is required to receive its value
        # when the object is created. If I think so, I put self.attribute = param_name
        # and add a new parameter at the header
        # I want to receive the image when I create the player, so the image param
        # has been added to the header
        self.image = image
        # Same for the side
        self.side = side
        # The player will start always at the lower level, so I set directly the
        # value, no parameter is needed then.
        self.vertical_pos = 0
````

````python
mario = Player("left_mario.png", "left")
luigi = Player("right_mario.png", "right")
# To see the values
print("The image of Mario is", mario.image)
print("The vertical position of Luigi is", luigi.vertical_pos)
# To change them
luigi.vertical_pos = 1
print("The vertical position of Luigi is", luigi.vertical_pos)
mario.image = "left_mario_taking_package.png"
print("The image of Mario is", mario.image)
# mario and luigi are called objects
````

# Types of variables in OOP

1. Attributes: store the information about the object. They are defined always at `__init__` and go always with `self.`
2. Parameters: appear on the header of the method and are used inside the method. In the `__init__` method they are used to give the initial value of the attributes (we will see that other methods usually have parameters too).
3. Local variables: temporal or auxiliar variables created inside a method (without `self.`). Once the method finishes they disappear and cannot be used anymore.

# Protecting attributes

How we do avoid that attributes are given wrong values?

Two ways:
1. Being nice, annotating what the right values are.
2. Being rude, throwing errors if the values are not right.

# Annotating

We can annotate the type of the parameters and we can also give information about them in the documentation.

This is only a recommendation, people creating objects of our class can totally ignore it.

We annotate in two different ways:

1. At the header of the method we put the required type for each of the parameters by writing `param_name : type`.
2. At the documentation we describe each parameter using `:param param_name: description`. Also `@param param_name: description`.

````python
# A class to represent a Date
class Date:
    """ This class represents dates
    Attributes
    :day : int. A value in the range 1 to 31, for some months it can be restricted
    to 28, 29 or 30
    :month : str. A string with the name of the month like in 'january'
    :year : int. A number not equal to 0 """
    def __init__(self, day: int, month: str, year: int):
        """ This method creates a Date by receiving all the information needed
            :param day: int. A value in the range 1 to 31, for some months it can be restricted
            to 28, 29 or 30
            :param month: str. A string with the name of the month like in 'january'
            :param year: int. A number not equal to 0
        """
        self.day = day
        self.month = month
        self.year = year
````

# Checking the values of the attributes

All OO languages have mechanisms to protect the attributes against wrong values. They allow you to create some code to check that the values are correct. In most languages we develop a method to check the value of the attribute (usually called `setter`) and another method to return the value of the attribute (usually called `getter`).

In Python, we have something similar but better: using `setters` and `properties`

To protect an attribute, we do the following:

1. Create a method that will receive a value, check that it is correct and if it is, it will assign the value to the attribute. If not, it will raise an error. The name of the method, must be the same of the attribute.



````python
# A class to represent a Date, with setter and property for year
class Date:
    """ This class represents dates. It has setters and properties to get sure
    the values of attributes are right.
    Attributes
    :day : int. A value in the range 1 to 31, for some months it can be restricted
    to 28, 29 or 30
    :month : str. A string with the name of the month like in 'january'
    :year : int. A number not equal to 0 """
    def __init__(self, day: int, month: str, year: int):
        """ This method creates a Date by receiving all the information needed
            :param day: int. A value in the range 1 to 31, for some months it can be restricted
            to 28, 29 or 30
            :param month: str. A string with the name of the month like in 'january'
            :param year: int. A number not equal to 0
        """
        self.day = day
        self.month = month
        self.year = year

    # This @property tells us that this is a magic method
    @property
    def year(self):
        """ This magic method must be created if we want the setter to work. If
        we don't create it before the setter, we will have an error. """
        # Here we use the name of the attribute with __ as in the setter, otherwise
        # it will not work
        return self.__year

    # With this @year.setter I am telling the interpreter that this is a magic
    # method that must be used every time I want to change the value of the year
    @year.setter
    def year(self, year: int):
        """ This method will be invoked every time we want to change the year
        :param year: a number not equal to 0
        """
        # We are checking first the type, we use better isinstance rather than type
        if not isinstance(year, int):
            raise TypeError("The year must be an integer")
        elif year == 0:
            raise ValueError("No year 0 allowed")
        else:
            # In the setter the name of the attribute must have __ before
            self.__year = year
````

````python
# A class to represent a Date, with setter and property for year, month and day
class Date:
    """ This class represents dates. It has setters and properties to get sure
    the values of attributes are right.
    Attributes
    :day : int. A value in the range 1 to 31, for some months it can be restricted
    to 28, 29 or 30
    :month : str. A string with the name of the month like in 'january'
    :year : int. A number not equal to 0 """
    def __init__(self, day: int, month: str, year: int):
        """ This method creates a Date by receiving all the information needed
            :param day: int. A value in the range 1 to 31, for some months it can be restricted
            to 28, 29 or 30
            :param month: str. A string with the name of the month like in 'january'
            :param year: int. A number not equal to 0
        """
        self.month = month
        self.year = year
        # We need to create the day after the month and the year because we
        # use year and month to check if the day is right in the day.setter
        self.day = day

    # This @property tells us that this is a magic method
    @property
    def year(self):
        """ This magic method must be created if we want the setter to work. If
        we don't create it before the setter, we will have an error. """
        # Here we use the name of the attribute with __ as in the setter, otherwise
        # it will not work
        return self.__year

    # With this @year.setter I am telling the interpreter that this is a magic
    # method that must be used every time I want to change the value of the year
    @year.setter
    def year(self, year: int):
        """ This method will be invoked every time we want to change the year
        :param year: a number not equal to 0
        """
        # We are checking first the type, we use better isinstance rather than type
        if not isinstance(year, int):
            raise TypeError("The year must be an integer")
        elif year == 0:
            raise ValueError("No year 0 allowed")
        else:
            # In the setter the name of the attribute must have __ before
            self.__year = year

    # With -> str I annotate that the method is returning a string. Not compulsory
    # but nice to be done.
    @property
    def month(self) -> str:
        return self.__month

    @month.setter
    def month(self, month: str):
        # A local variable with a dictionary with the names of the months and
        # their number of days
        months_days = {"january": 31, "february": 28, "march" : 31, "april": 30}
        if not isinstance(month, str):
            raise TypeError("The month must be a string")
        elif month.lower() not in months_days:
            raise ValueError("The valid names for months are " + str(tuple(months_days.keys())))
        else:
            self.__month = month.lower()

    @property
    def day(self) -> int:
        return self.__day

    @day.setter
    def day(self, day: int):
        if not isinstance(day, int):
            raise TypeError("The day must be an int")
        else:
            # We cannot reuse the variable we created in month setter so we declare
            # it again here. In a couple of weeks we will see a way to avoid that
            months_days = {"january": 31, "february": 28, "march" : 31, "april": 30}
            # We check if we are in a leap year, to change the number of days of
            # february. Notice we use self.year and not self.__year as we are not
            # in the year setter but in the day setter
            if self.year % 4 == 0 and (self.year % 100 != 0 or self.year % 400 == 0):
                months_days["february"] = 29
            # Looking for the maximum number of days for the self.month
            max_days = 0
            for key in months_days:
                # Notice we use self.month and not self.__month as we did with self.year
                if key == self.month:
                    max_days = months_days[key]
            if day < 1 or day > max_days:
                raise ValueError("The maximum number of days for " + str(self.month)
                + " is " + str(max_days) + " for year " + str(self.year))
            else:
                self.__day = day
````

# More magic methods

In addition to `init` and properties/setters our classes usually contain more magic methods.

The following ones are usually implemented:

1. `__str__`: it returns the string representation of an object when it is printed.
2. `__repr__`: it returns the string representation of an object when it is interpreted. Usually same as str.
3. `__eq__`: it is used to compare two objects. It is invoked when we do `object1 == object2`
4. `__lt__` is invoked when we do `object1 < object2`
5. `__le__` is invoked when we do `object1 <= object2`
6. We could also implement `__gt__`, `__ge__` or `__ne__`  but most times is not needed as they are the inverse of the previous ones

````python
# We include the __str__,  __repr__, __eq__, __lt__ and __leq__ methods
class Date:
    """ This class represents dates. It has setters and properties to get sure
    the values of attributes are right.
    Attributes
    :day : int. A value in the range 1 to 31, for some months it can be restricted
    to 28, 29 or 30
    :month : str. A string with the name of the month like in 'january'
    :year : int. A number not equal to 0 """
    def __init__(self, day: int, month: str, year: int):
        """ This method creates a Date by receiving all the information needed
            :param day: int. A value in the range 1 to 31, for some months it can be restricted
            to 28, 29 or 30
            :param month: str. A string with the name of the month like in 'january'
            :param year: int. A number not equal to 0
        """
        self.month = month
        self.year = year
        # We need to create the day after the month and the year because we
        # use year and month to check if the day is right in the day.setter
        self.day = day

    # This @property tells us that this is a magic method
    @property
    def year(self):
        """ This magic method must be created if we want the setter to work. If
        we don't create it before the setter, we will have an error. """
        # Here we use the name of the attribute with __ as in the setter, otherwise
        # it will not work
        return self.__year

    # With this @year.setter I am telling the interpreter that this is a magic
    # method that must be used every time I want to change the value of the year
    @year.setter
    def year(self, year: int):
        """ This method will be invoked every time we want to change the year
        :param year: a number not equal to 0
        """
        # We are checking first the type, we use better isinstance rather than type
        if not isinstance(year, int):
            raise TypeError("The year must be an integer")
        elif year == 0:
            raise ValueError("No year 0 allowed")
        else:
            # In the setter the name of the attribute must have __ before
            self.__year = year

    # With -> str I annotate that the method is returning a string. Not compulsory
    # but nice to be done.
    @property
    def month(self) -> str:
        return self.__month

    @month.setter
    def month(self, month: str):
        # A local variable with a dictionary with the names of the months and
        # their number of days
        months_days = {"january": 31, "february": 28, "march" : 31, "april": 30}
        if not isinstance(month, str):
            raise TypeError("The month must be a string")
        elif month.lower() not in months_days:
            raise ValueError("The valid names for months are " + str(tuple(months_days.keys())))
        else:
            self.__month = month.lower()

    @property
    def day(self) -> int:
        return self.__day

    @day.setter
    def day(self, day: int):
        if not isinstance(day, int):
            raise TypeError("The day must be an int")
        else:
            # We cannot reuse the variable we created in month setter so we declare
            # it again here. In a couple of weeks we will see a way to avoid that
            months_days = {"january": 31, "february": 28, "march" : 31, "april": 30}
            # We check if we are in a leap year, to change the number of days of
            # february. Notice we use self.year and not self.__year as we are not
            # in the year setter but in the day setter
            if self.year % 4 == 0 and (self.year % 100 != 0 or self.year % 400 == 0):
                months_days["february"] = 29
            # Looking for the maximum number of days for the self.month
            max_days = 0
            for key in months_days:
                # Notice we use self.month and not self.__month as we did with self.year
                if key == self.month:
                    max_days = months_days[key]
            if day < 1 or day > max_days:
                raise ValueError("The maximum number of days for " + str(self.month)
                + " is " + str(max_days) + " for year " + str(self.year))
            else:
                self.__day = day

    def month_number(self) -> int:
        """ This method returns the number of the month of this date
        :return : 1 for january, 2 for february and so on
        """
        # A local variable to store the names of the months
        months_names = ("january", "february", "march", "april")
        for pos in range(len(months_names)):
            if self.month == months_names[pos]:
                number = pos + 1
        return number

    # No more parameters can be added
    def __str__(self) -> str:
        """ This method is invoked when we print the object. It must return a
        string that is what we want to be printed"""
        # Local variable
        aux = "th"
        if self.day == 1 or self.day == 11 or self.day == 21 or self.day == 31:
            aux = "st"
        elif self.day == 2 or self.day == 12 or self.day == 22:
            aux = "nd"
        elif self.day == 3 or self.day == 13 or self.day == 23:
            aux = "rd"
        # Another local variable
        month_name = self.month[0].upper() + self.month[1:]
        return month_name + " " + str(self.day) + aux + ", " + str(self.year)

    # Again, it must be named like this for it to work
    def __repr__(self) -> str:
        # We usually just invoke the str
        return self.__str__()

    # I need to receive the object I want this to be compared to
    def __eq__(self, another) -> bool:
        if not isinstance(another, Date):
            # By convention if the object is not of this class, we return notImplemented
            return NotImplemented
        else:
            return (self.day == another.day and self.month == another.month
                    and self.year == another.year)

    def __lt__(self, another) -> bool:
        """ This method is invoked when we do object1 < object2 """
        if not isinstance(another, Date):
            # By convention if the object is not of this class, we return notImplemented
            return NotImplemented
        else:
            result = False
            if self.year < another.year:
                result = True
            elif self.year == another.year and self.month_number() < another.month_number():
                result = True
            elif (self.year == another.year and self.month_number() == another.month_number()
                  and self.day < another.day):
                result = True
            return result
    def __le__(self, another) -> bool:
        """ This method is invoked when we do object1 <= object2 """
        return self == another or self < another
````

# Composition

There are two ways to create a class using another class: composition and inheritance

Composition: an object is an attribute of another object


````python
# A class to show composition
class Student:
    def __init__(self, name: str, surname: str, birthday: Date):
        self.name = name
        self.surname = surname
        # This attribute is an object of the Date class
        self.birthday = birthday

    @property
    def name(self) -> str:
        return self.__name

    @property
    def surname(self) -> str:
        return self.__surname

    @property
    def birthday(self) -> Date:
        return self.__birthday

    @name.setter
    def name(self, name: str):
        if not isinstance(name, str):
            raise TypeError("The name must be a str")
        else:
            self.__name = name

    @surname.setter
    def surname(self, surname: str):
        if not isinstance(surname, str):
            raise TypeError("The name must be a str")
        else:
            self.__surname = surname

    @birthday.setter
    def birthday(self, birthday: Date):
        if not isinstance(birthday, Date):
            raise TypeError("The birthday must be a Date")
        else:
            # The Date must be correct as no wrong incorrect object can be created
            self.__birthday = birthday

    def __str__(self) -> str:
        return self.name + " " + self.surname + " born on " + str(self.birthday)
````

````python
# Two ways to create a composite object
# 1) We first create the attribute
bd = Date(1, "january", 2000)
st = Student("Pepe", "Perez", bd)
print(st)
# 2) we create the attribute on the fly (best way)
st2 = Student("Luisa", "Gomez", Date(2, "march", 2000))
print(st2)
# Changing the day Pepe was born
st.birthday.day = 12
print(st)
````

````python
# An example of a class with an attibute that is a list of objects
class Classroom:
    def __init__(self, students: list[Student]):
        self.students  = students

    @property
    def students(self) -> list[Student]:
        return self.__students

    @students.setter
    def students(self, students):
        if not isinstance(students, list):
            raise TypeError ("The students must be a list")
        else:
            for element in students:
                if not isinstance(element, Student):
                    raise ValueError("All the students must be of type Student")
        # If we are here, everything  went well
        self.__students = students

    def __str__(self) -> str:
        result = "The class is:\n"
        # We add every student to the str
        for student in self.students:
            result += str(student) + "\n"
        # Once all are added, we return them
        return result

````

# Encapsulation: information hiding

Encapsulation is one on the three main characteristcics of OO: it means to put together the information (attributes) and the things that an object can do (methods)

Information hiding: We keep some information private inside the class or we allow to read but no to change some information.

Read only attributes: attributes whose value can be read outside the class but not changed. Two types:

1. Calculated attributes: its value can be calculated using the values of other attributes. We create a property for it, but not a setter. The calculation will be performed inside the property.
2. Non calculated ones: We want to keep them read only but they cannot be calculated. Their initial value must be set in the init. They must be created in the init but with `__`. Then, again, we will create a property, but not a setter.

# Private attributes

A private attribute is an attribute that cannot be changed nor read outside the class. Is used to store information that we need to use inside the class but we don't want to be seen outside. We create the attribute in the init with `__`. No property or setter for it.

`__` means the attribute is private and cannot be seen outside the class.

````python
# Adding a private attribute
class Date:
    """ This class represents dates. It has setters and properties to get sure
    the values of attributes are right.
    Attributes
    :day : int. A value in the range 1 to 31, for some months it can be restricted
    to 28, 29 or 30
    :month : str. A string with the name of the month like in 'january'
    :year : int. A number not equal to 0 """
    def __init__(self, day: int, month: str, year: int):
        """ This method creates a Date by receiving all the information needed
            :param day: int. A value in the range 1 to 31, for some months it can be restricted
            to 28, 29 or 30
            :param month: str. A string with the name of the month like in 'january'
            :param year: int. A number not equal to 0
        """
         # This is a private attribute, it can be used by any method of the class
        # but is not accesible outside
        self.__months_days = {"january": 31, "february": 28, "march" : 31, "april": 30}
        self.month = month
        self.year = year
        # We need to create the day after the month and the year because we
        # use year and month to check if the day is right in the day.setter
        self.day = day

    # This @property tells us that this is a magic method
    @property
    def year(self):
        """ This magic method must be created if we want the setter to work. If
        we don't create it before the setter, we will have an error. """
        # Here we use the name of the attribute with __ as in the setter, otherwise
        # it will not work
        return self.__year

    # With this @year.setter I am telling the interpreter that this is a magic
    # method that must be used every time I want to change the value of the year
    @year.setter
    def year(self, year: int):
        """ This method will be invoked every time we want to change the year
        :param year: a number not equal to 0
        """
        # We are checking first the type, we use better isinstance rather than type
        if not isinstance(year, int):
            raise TypeError("The year must be an integer")
        elif year == 0:
            raise ValueError("No year 0 allowed")
        else:
            # In the setter the name of the attribute must have __ before
            self.__year = year
            # I need to change also the days of February if the year is leap
            if self.leap_year:
                self.__months_days['february'] = 29
            else:
                self.__months_days['february'] = 28

    # This property has no setter so it is a read only attribute
    @property
    def leap_year(self) -> bool:
        return self.year % 4 == 0 and (self.year % 100 != 0 or self.year % 400 == 0)

    # With -> str I annotate that the method is returning a string. Not compulsory
    # but nice to be done.
    @property
    def month(self) -> str:
        return self.__month

    @month.setter
    def month(self, month: str):
        if not isinstance(month, str):
            raise TypeError("The month must be a string")
        elif month.lower() not in self.__months_days:
            raise ValueError("The valid names for months are " +
                             str(tuple(self.__months_days.keys())))
        else:
            self.__month = month.lower()

    @property
    def day(self) -> int:
        return self.__day

    @day.setter
    def day(self, day: int):
        if not isinstance(day, int):
            raise TypeError("The day must be an int")
        else:
            # Looking for the maximum number of days for the self.month
            max_days = 0
            for key in self.__months_days:
                # Notice we use self.month and not self.__month as we did with self.year
                if key == self.month:
                    max_days = self.__months_days[key]
            if day < 1 or day > max_days:
                raise ValueError("The maximum number of days for " + str(self.month)
                + " is " + str(max_days) + " for year " + str(self.year))
            else:
                self.__day = day

    def month_number(self) -> int:
        """ This method returns the number of the month of this date
        :return : 1 for january, 2 for february and so on
        """
        # A local variable to store the names of the months
        months_names = ("january", "february", "march", "april")
        for pos in range(len(months_names)):
            if self.month == months_names[pos]:
                number = pos + 1
        return number

    # No more parameters can be added
    def __str__(self) -> str:
        """ This method is invoked when we print the object. It must return a
        string that is what we want to be printed"""
        # Local variable
        aux = "th"
        if self.day == 1 or self.day == 21 or self.day == 31:
            aux = "st"
        elif self.day == 2 or self.day == 22:
            aux = "nd"
        elif self.day == 3 or self.day == 23:
            aux = "rd"
        # Another local variable
        month_name = self.month[0].upper() + self.month[1:]
        # Another local one
        leap = " (not leap)"
        if self.leap_year:
            leap = " (leap)"
        return month_name + " " + str(self.day) + aux + ", " + str(self.year) + leap

    # Again, it must be named like this for it to work
    def __repr__(self) -> str:
        # We usually just invoke the str
        return self.__str__()

    # I need to receive the object I want this to be compared to
    def __eq__(self, another) -> bool:
        if not isinstance(another, Date):
            # By convention if the object is not of this class, we return notImplemented
            return NotImplemented
        else:
            return (self.day == another.day and self.month == another.month
                    and self.year == another.year)

    def __lt__(self, another) -> bool:
        """ This method is invoked when we do object1 < object2 """
        if not isinstance(another, Date):
            # By convention if the object is not of this class, we return notImplemented
            return NotImplemented
        else:
            result = False
            if self.year < another.year:
                result = True
            elif self.year == another.year and self.month_number() < another.month_number():
                result = True
            elif (self.year == another.year and self.month_number() == another.month_number()
                  and self.day < another.day):
                result = True
            return result
    def __le__(self, another) -> bool:
        """ This method is invoked when we do object1 <= object2 """
        return self == another or self < another
````

# When do we use `__`?

1. In the setters and the properties always. Actually what we are doing is to store the information in an equivalent private attribute and use the property and the setter to work with it.
2. In the init for a non-calculated read only attribute. Then we use a property to show its value.
3. In the init for a private attribute. No property or setter is created for it.

# Private methods

Methods whose name starts by `__`. They can be used inside the class but not outside it. Usually used for auxiliary methods.

````python
# Adding a private auxiliary method in Student to check the id
class Student:
    def __init__(self, name: str, surname: str, birthday: Date, st_id: str):
        self.name = name
        self.surname = surname
        # This attribute is an object of the Date class
        self.birthday = birthday
        # As we don't have a setter for this attribute, the checking must be done
        # here, we use a private method
        if self.__check_id(st_id):
            self.__student_id = st_id

    @property
    def name(self) -> str:
        return self.__name

    @property
    def surname(self) -> str:
        return self.__surname

    @property
    def birthday(self) -> Date:
        return self.__birthday

    #This is a read only attribute as there is no setter for it
    @property
    def student_id(self) -> str:
        return self.__student_id

    @name.setter
    def name(self, name: str):
        if not isinstance(name, str):
            raise TypeError("The name must be a str")
        else:
            self.__name = name

    @surname.setter
    def surname(self, surname: str):
        if not isinstance(surname, str):
            raise TypeError("The name must be a str")
        else:
            self.__surname = surname

    @birthday.setter
    def birthday(self, birthday: Date):
        if not isinstance(birthday, Date):
            raise TypeError("The birthday must be a Date")
        else:
            # The Date must be correct as no wrong incorrect object can be created
            self.__birthday = birthday

    def __str__(self) -> str:
        return (self.name + " " + self.surname + " with id "  + self.student_id
                + ", born on " + str(self.birthday))

    def __check_id(self, id: str) -> bool:
        """ Private method that checks the id is right and returns if it is
        :param id:  the id
        :return: True if the id is right"""
        if not isinstance(id, str):
            raise  TypeError("The id  must be a string")
        elif len(id) != 9:
            raise ValueError("The id must have 9 characters")
        else:
            return True
````

# Inheritance

Second way to create a clas using another class, after composition.

In composition a class has an attribute that is an object of another class.

In inheritance, a class is an extension of another class. The new class will have all the attributes and mehtods of the other class, plus more attributes and methods.


````python
# A class that will be used to inherit from
class Product:
    def __init__(self, name: str, price: float, buy_price: float):
        self.name = name
        self.price = price
        # A read only non-calculated attribute
        self.__buy_price = buy_price

    # No property/setter for name as anything can be a Product name

    @property
    def price(self) -> float:
        return self.__price

    @price.setter
    def price(self, price: float):
        if not (isinstance(price, float) or isinstance(price, int)):
            raise TypeError("The price must be a number")
        elif price < 0:
            raise ValueError("A negative price, really?")
        else:
            self.__price = price

    @property
    def buy_price(self) -> float:
        return self.__buy_price

    def __str__(self) -> str:
        return self.name + ": " + str(self.price)+"€"

    def __eq__(self, another) -> bool:
        """ Two products will be the same if the have the same name, despite the
        price """
        if not isinstance(another, Product):
            return NotImplemented
        else:
            return self.name == another.name

    def profit(self) -> float:
        return self.price - self.buy_price

    def sale(self):
        """ Price reduced by a 50%"""
        self.price = round(self.price / 2, 2)
````

````python
p = Product("Computer", 1500, 1000)
print(p)
print("The buy price was", p.buy_price)
print("The profit a the current price is", p.profit())
p.sale()
print(p)
print("The profit a the current price is", p.profit())
````

````python
# A FreshProduct class that only has an extra method (with no inheritance)
class FreshProduct:
    def __init__(self, name: str, price: float, buy_price: float):
        self.name = name
        self.price = price
        # A read only non-calculated attribute
        self.__buy_price = buy_price

    # No property/setter for name as anything can be a Product name

    @property
    def price(self) -> float:
        return self.__price

    @price.setter
    def price(self, price: float):
        if not (isinstance(price, float) or isinstance(price, int)):
            raise TypeError("The price must be a number")
        elif price < 0:
            raise ValueError("A negative price, really?")
        else:
            self.__price = price

    @property
    def buy_price(self) -> float:
        return self.__buy_price

    def __str__(self) -> str:
        return self.name + ": " + str(self.price)+"€"

    def __eq__(self, another) -> bool:
        """ Two products will be the same if the have the same name, despite the
        price """
        if not isinstance(another, Product):
            return NotImplemented
        else:
            return self.name == another.name

    def profit(self) -> float:
        return self.price - self.buy_price

    def sale(self):
        """ Price reduced by a 50%"""
        self.price = round(self.price / 2, 2)

    # The new method
    def rotten_sale(self):
        """ When the fresh product is close to be rotten, we sell it a 0.1€"""
        self.price = 0.1
````

If we just want to add new methods, it is as easy as to just add them in the new class.

But if we want to add new attributes, we need to rewrite the init.

Product is known as the mother class or the superclass

FreshProduct is known as the child class or the subclass

````python
# Adding a new attribute to the child class
class FreshProduct(Product):
    # We need to copy the init again
    def __init__(self, name: str, price: float, buy_price: float, expiration: Date):
        # This means, execute the init of the mother class
        super().__init__(name, price, buy_price)
        self.expiration = expiration

    @property
    def expiration(self) -> Date:
        return self.__expiration

    @expiration.setter
    def expiration(self, expiration: Date):
        if not isinstance(expiration, Date):
            raise TypeError("The expiration must be a Date")
        else:
            self.__expiration = expiration

    # We rewrite the str to show also the expiration date
    def __str__(self) -> str:
        return super().__str__() + " expires on " + str(self.expiration)

    def __eq__(self, another) -> bool:
        if not isinstance(another, FreshProduct):
            return NotImplemented
        else:
            # We also rely on the mother eq to compare the Product part of this object
            return super().__eq__(another) and self.expiration == another.expiration

    # The new method
    def rotten_sale(self):
        """ When the fresh product is close to be rotten, we sell it a 0.1€"""
        self.price = 0.1
````

````python
fp2 = FreshProduct("Oranges", 1, 0.25, Date(12, "january", 2026))
print(fp2)
````

````python
# A class that inherits from FreshProduct (and also from Product)
class FrozenProduct(FreshProduct):
    def __init__(self, name: str, price: float, buy_price: float, exp: Date):
        super().__init__(name, price, buy_price, exp)
        # By default frozen products will start frozen
        self.__frozen = True

    @property
    def frozen(self) -> bool:
        return self.__frozen

    def unfreeze(self):
        """ We only allow the value of the attribute to go from True to False"""
        self.__frozen = False

    def __str__(self) -> str:
        aux  = " (frozen)"
        if not self.__frozen:
            aux = " (not frozen)"
        return super().__str__() + aux
````

````python
class Mother():
    def __init__(self, private: int):
        if not isinstance(private, int):
            raise TypeError("The private attribute must be an int")
        else:
            self.__private = private

    def __str__(self) -> str:
        result = "The value of the attribute is "
        if self.__private > 0:
            result += "positive"
        else:
            result += "negative"
        return result
````

````python
m = Mother(43)
m2 = Mother(-1)
print(m)
print(m2)
````

````python
# A child class that uses a protected attribute
class Child(Mother):
    def __init__(self, private: int, public: int):
        super().__init__(private)
        self.public = public

    @property
    def public(self) -> int:
        return self.__public

    @public.setter
    def public(self, value: int):
        if not isinstance(value, int):
            raise TypeError("The public attribute must be a int")
        else:
            self.__public = value

    def __str__(self) -> str:
        aux = " and greater than "
        if self.public >= self._private:
            aux = " and less than or equal to "
        return super().__str__() + aux + str(self.public)
````

# Optional parameters

````python
# Now all the parameters of the init are optional
class Date:
    """ This class represents dates. It has setters and properties to get sure
    the values of attributes are right.
    Attributes
    :day : int. A value in the range 1 to 31, for some months it can be restricted
    to 28, 29 or 30
    :month : str. A string with the name of the month like in 'january'
    :year : int. A number not equal to 0 """
    def __init__(self, day: int = 1, month: str = "january", year: int = 2025):
        """ This method creates a Date by receiving all the information needed
            :param day: int. A value in the range 1 to 31, for some months it can be restricted
            to 28, 29 or 30
            :param month: str. A string with the name of the month like in 'january'
            :param year: int. A number not equal to 0
        """
         # This is a private attribute, it can be used by any method of the class
        # but is not accesible outside
        self.__months_days = {"january": 31, "february": 28, "march" : 31, "april": 30}
        self.month = month
        self.year = year
        # We need to create the day after the month and the year because we
        # use year and month to check if the day is right in the day.setter
        self.day = day

    # This @property tells us that this is a magic method
    @property
    def year(self):
        """ This magic method must be created if we want the setter to work. If
        we don't create it before the setter, we will have an error. """
        # Here we use the name of the attribute with __ as in the setter, otherwise
        # it will not work
        return self.__year

    # With this @year.setter I am telling the interpreter that this is a magic
    # method that must be used every time I want to change the value of the year
    @year.setter
    def year(self, year: int):
        """ This method will be invoked every time we want to change the year
        :param year: a number not equal to 0
        """
        # We are checking first the type, we use better isinstance rather than type
        if not isinstance(year, int):
            raise TypeError("The year must be an integer")
        elif year == 0:
            raise ValueError("No year 0 allowed")
        else:
            # In the setter the name of the attribute must have __ before
            self.__year = year
            # I need to change also the days of February if the year is leap
            if self.leap_year:
                self.__months_days['february'] = 29
            else:
                self.__months_days['february'] = 28

    # This property has no setter so it is a read only attribute
    @property
    def leap_year(self) -> bool:
        return self.year % 4 == 0 and (self.year % 100 != 0 or self.year % 400 == 0)

    # With -> str I annotate that the method is returning a string. Not compulsory
    # but nice to be done.
    @property
    def month(self) -> str:
        return self.__month

    @month.setter
    def month(self, month: str):
        if not isinstance(month, str):
            raise TypeError("The month must be a string")
        elif month.lower() not in self.__months_days:
            raise ValueError("The valid names for months are " +
                             str(tuple(self.__months_days.keys())))
        else:
            self.__month = month.lower()

    @property
    def day(self) -> int:
        return self.__day

    @day.setter
    def day(self, day: int):
        if not isinstance(day, int):
            raise TypeError("The day must be an int")
        else:
            # Looking for the maximum number of days for the self.month
            max_days = 0
            for key in self.__months_days:
                # Notice we use self.month and not self.__month as we did with self.year
                if key == self.month:
                    max_days = self.__months_days[key]
            if day < 1 or day > max_days:
                raise ValueError("The maximum number of days for " + str(self.month)
                + " is " + str(max_days) + " for year " + str(self.year))
            else:
                self.__day = day

    def month_number(self) -> int:
        """ This method returns the number of the month of this date
        :return : 1 for january, 2 for february and so on
        """
        # A local variable to store the names of the months
        months_names = ("january", "february", "march", "april")
        for pos in range(len(months_names)):
            if self.month == months_names[pos]:
                number = pos + 1
        return number

    # No more parameters can be added
    def __str__(self) -> str:
        """ This method is invoked when we print the object. It must return a
        string that is what we want to be printed"""
        # Local variable
        aux = "th"
        if self.day == 1 or self.day == 21 or self.day == 31:
            aux = "st"
        elif self.day == 2 or self.day == 22:
            aux = "nd"
        elif self.day == 3 or self.day == 23:
            aux = "rd"
        # Another local variable
        month_name = self.month[0].upper() + self.month[1:]
        # Another local one
        leap = " (not leap)"
        if self.leap_year:
            leap = " (leap)"
        return month_name + " " + str(self.day) + aux + ", " + str(self.year) + leap

    # Again, it must be named like this for it to work
    def __repr__(self) -> str:
        # We usually just invoke the str
        return self.__str__()

    # I need to receive the object I want this to be compared to
    def __eq__(self, another) -> bool:
        if not isinstance(another, Date):
            # By convention if the object is not of this class, we return notImplemented
            return NotImplemented
        else:
            return (self.day == another.day and self.month == another.month
                    and self.year == another.year)

    def __lt__(self, another) -> bool:
        """ This method is invoked when we do object1 < object2 """
        if not isinstance(another, Date):
            # By convention if the object is not of this class, we return notImplemented
            return NotImplemented
        else:
            result = False
            if self.year < another.year:
                result = True
            elif self.year == another.year and self.month_number() < another.month_number():
                result = True
            elif (self.year == another.year and self.month_number() == another.month_number()
                  and self.day < another.day):
                result = True
            return result
    def __le__(self, another) -> bool:
        """ This method is invoked when we do object1 <= object2 """
        return self == another or self < another
````

# Passing by value and passing by reference

Pass by reference: any change done to the parameter inside the method is actually done to the variable used to invoke the method. Objects, dictionaries and lists pass by reference.

Pass by value: any change done to the parameters inside the method is not done to the variable used to invoke the method. The value of the variable is copied to the parameter and we work with the copy. Int, float, str, bool, tuples... pass by value.

In C++ you can choose.

````python
from IPython.core.application import ProfileAwareConfigLoader
class Soldier:
    def __init__(self, power: int):
        self.power = power

    @property
    def power(self) -> int:
        return self.__power

    @power.setter
    def power(self, value: int):
        if not isinstance(value, int):
            raise TypeError("The power must be int")
        elif value < 0:
            raise ValueError("the power cannot we negative")
        else:
            self.__power = value

    def __str__(self) -> str:
        return "Power of the soldier: " + str(self.power)

    def fight(self, another):
        # We should check another is also a soldier
        if self.power >= another.power:
            self.power = self.power - another.power
            another.power = 0
        else:
            another.power -= self.power
            self.power = 0

    def fight2(self, power: int):
        if self.power >= power:
            self.power = self.power - power
            power = 0
        else:
            power -= self.power
            self.power = 0
````

````python
s1 = Soldier(10)
s2 = Soldier(7)
print(s1)
print(s2)
# The power of s2 changes inside the method, and also affects s2
s1.fight(s2)
print(s1)
print(s2)
````
