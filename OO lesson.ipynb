# Object oriented programming

Once we have identified the important concepts and we have create a class diagram (UML), describing the attributes (data to store) and the methods (things they can do) of every class, we can start programming them.

We will start a new file with this:
```
class NameOfTheClass:
    """ Description of the class, explaining what it is doing"""
    Code of the class
```

Rules: name of the class, we use upper camel case

A class in each file. The name of the file should be the name of the class starting by lower case: `nameOfTheClass.py`

The string in """ """ is called *docstring* and it is optional, but compulsory this year.

````python
# Our initial class, nothing is done yet
class Date:
    """ This class represents a Date to be used in my programs. It is our first
    example of OOP"""
    def __init__(self, parameter1, parameter2, parameter3):
        """ The constructor of the class Date. It initializes the attributes
        day, month and year with the values received as parameters"""
        self.attribute1 = parameter1
        self.attribute2 = parameter2
        self.attribute3 = parameter3
        
````

# Methods

All the code inside a class must be inside a method, no code can be alone

```
def name_of_the_method(self, param2, param3, ...):
    """ Method documentation """
    instruction1
    instruction2
```

So, our classes will always be like:

```
class NameOfTheClass:
    """ Description of the class, explaining what it is doing"""

    def method1(self, param2, param3, ...):
        """ Method documentation """
        instruction1
        instruction2Code of the class

    def method2(self, param2, param3, ...):
        """ Method documentation """
        instruction1
        instruction2Code of the class

    def method3(self, param2, param3, ...):
        """ Method documentation """
        instruction1
        instruction2
```

Parts of a method:

1. Header: contains the name and the parameters (the information needed to run the method). There will be always 1 parameter, the first one, which is called `self`. The remaining ones are just variable names.
2. Documentation: explains what the method is doing
3. Body of the method: the real code.


````python
# A class to represent character for the final project, it only contains the attributes
class character:
    """ This class represents the characters """
    def __init__(self, image: str, side: str):
        """ This is the magic method we must use to declare the attributes of our
        objects"""
        # My attributes must always start by self.
        self.image = image
        self.side = side
        # I assume character will always start on the lower part of the screen so
        # I don't need a parameter for this attribute
        self.vertical_pos = 0
````

### Outside of the class you can give the values wanted to the attributes

# Type of variables is OOP

1. Attributes: store the information about the object. They are defined always at `__innit__` adn go always with `self`
2. Parameters: appear in the header of the method and are used inside the method. In the `__innit__` method theyre used to give the initial value of the attributes( we will see that other methods usually have aprameters too)
3. Local variables: temporal or auxiliar variables created inside a method (without `self`). Once the method finishes they disappear and cannot be used anymore

# Protecting attributes

How do we avoid that attributes are given wrong values?

Two ways:
1. Explaining what the correct values are
2. Sending errors if the values are not correct

# Annotating

We can annotate the type of parameters and we can also give information about them in the documentation.

This is only a recommendation, people creating objects of our classes can totally ignore it. 

We annotate in two different ways:

1. At the header of the method we put the required type for each of the parameters bu writing `param_name : type`.
2. At the documentation we describe each parameter using `:param param_name: description`. Also `@param param_name: description


````python
#A class to represent a Date with setters and getters to check if the attributes are correct
class Date:
    """ This class represents dates
    Attributes
    :day : int. A value in the range 1-31, for some months it can be restricted to 30 or 28/29
    :month : str. The name of the month, e.g., "January", "February", etc.
    :year : int. A number not equal to 0
    """
    def __init__(self, day: int, month: str, year: int):
        """ This method creates a Date by receiving all the information needed
        :param day : int. A value in the range 1-31, for some months it can be restricted to 30 or 28/29
        :param month : str. The name of the month, e.g., "January", "February", etc.
        :param year : int. A number not equal to 0
        """
        self.day = day
        self.month = month
        self.year = year

    @property
    def year(self):
        """ This magic method must be created if we want to use a setter for the year
        """
        #Here we put the name of the attribute with two underscores
        return self.__year

    #This property tells the interpreter that this is a magic method to set the year
    @year.setter
    def year(self, year: int):
        """ This method will be envolved every time that we want to change the year
        """
        if not isinstance(year, int):
            raise TypeError("The year must be an integer")
        elif year == 0:
            raise ValueError("The year cannot be 0")
        else:
            #In the setter, the name of the attribute must be different (__name)
            self.__year = year
````

````python
date1 = Date(27, "May", 2007)
print(date1.year)
````

# Checking the values of the attributes

All languages have mechanisms to protect the attributes against wrong values. They allow you to create some code to check that the values are valid.
In most languages we develop a method to check the value of the attribute (usually called `setter`) and another method to return the value of the attribute (usually called `getter`)

In Python we have something similar but better -> using `setters` and `properties`

To protect an attribute we do the following

1. Create a method that will receive a value, check that it is correct and if it is, it will assign the value to the attribute. If not, it will raise an error.
The name of the method must be the same of the attribute

````python
#A class to represent a Date with setters and getters to check if the attributes are correct
class Date:
    """ This class represents dates
    Attributes
    :day: int. A value in the range 1-31, for some months it can be restricted to 30 or 28/29
    :month : str. The name of the month, e.g., "January", "February", etc.
    :year : int. A number not equal to 0
    """
    def __init__(self, day: int, month: str, year: int):
        """ This method creates a Date by receiving all the information needed
        :param day: int. A value in the range 1-31, for some months it can be restricted to 30 or 28/29
        :param month : str. The name of the month, e.g., "January", "February", etc.
        :param year : int. A number not equal to 0
        """
        
        self.year = year
        self.month = month
        self.day = day

    #Tells that this is a magic method to get the month
    @property
    def month(self):
        """ This magic method must be created if we want to use a setter for the month
        """
        #Here we put the name of the attribute with two underscores
        return self.__month
    
    #This property tells the interpreter that this is a magic method to set the month
    @month.setter
    def month(self, month: str):
        """ This method will be envolved every time that we want to change the month
        """
        month_days = {"january":31, "february":28, "march":31, "april":30, "may":31, "june":30,
                      "july":31, "august":31, "september":30, "october":31, "november":30, "december":31}
        if not isinstance(month, str):
            raise TypeError("The month must be a string")
        elif month.lower() not in month_days:
            raise ValueError("The month must be" + str(tuple(month_days.keys())))
        else:
            #In the setter, the name of the attribute must be different (__name). ChatGPT uses only one underscore, they'll know we have used it
            self.__month = month

    @property
    def day(self):
        """ This magic method must be created if we want to use a setter for the day
        """
        #Here we put the name of the attribute with two underscores
        return self.__day
    @day.setter
    def day(self, day: int):
        """ This method will be envolved every time that we want to change the day
        """
        #You have to declare month_days again because its another method
        if not isinstance(day, int):
            raise TypeError("The day must be an integer")
        else:
            month_days = {"january":31, "february":28, "march":31, "april":30, "may":31, "june":30,
                        "july":31, "august":31, "september":30, "october":31, "november":30, "december":31}
            if self.year % 4 == 0 and (self.year % 100 != 0 or self.year % 400 == 0):
                month_days["february"] = 29
            max_days = 0
            for key in month_days:
                if key == self.month.lower():
                    max_days = month_days[key]
            if day < 1 or day > max_days:
                raise ValueError(f"The maximum number of days of" + str(self.month) + " is " + str(max_days))
            else:
                #In the setter, the name of the attribute must be different (__name)
                self.__day = day
                
````

````python
import random
class Package:
    """A class to represent a package, packages can start at any belt, we decide
    when creating the object. Direction will be random"""
    def __init__(self, y: int):
        """ The init method. We have decided that every package
        will have the same values for the attributes when we create
        them, except for the y and the direction which will be randomly selected.
        So no parameters in init method, just the self and y """
        # number is a local variable, it cannot be used outside the method
        # as we want it only to give value to direction, we don't put self
        # before it, that way we minimize the number of attributes
        number = random.randint(0,1)
        if number == 0:
            self.direction = "left"
        else:
            self.direction = "right"
        self.at_truck = False
        self.x = 0
        self.y = y
        # The image depends on the level, we assume 3 levels only
        if self.y == 0:
            self.image = "empty_package.png"
        elif self.y == 1:
            self.image = "level1_package.png"
        elif self.y == 2:
            self.image = "level2._package.png"
    
    @property
    def direction(self):
        return self.__direction
    
    @direction.setter
    def direction(self, direction: str):
        directions = ["left", "right"]
        if direction.lower() not in directions:
            raise ValueError("Direction must be 'left' or 'right'")
        else:
            self.__direction = direction.lower()
    
    @property
    def y(self):
        return self.__y
    
    @y.setter
    def y(self, y: int):
        if not isinstance(y, int):
            raise TypeError("The y must be an integer")
        elif y < 0:
            raise ValueError("The y must be greater or equal than 0")
        else:
            self.__y = y
````
