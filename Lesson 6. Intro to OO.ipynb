# Intro to OO (object orientation)
* With our current knowledge we can create simple programs. But when dealing with complex projects, particularly those that have to be performed by more than one person, we need programming methodologies to simplify collaboration and foster code reusing.

* When facing a big programming challenge we need to decompose it into smaller pieces that are easier to perform and that can be distributed among several programmers.

* Usually, what we do is to divide the program into several modules (a module is a file containing part of the program), each of them with part of the data and the instructions.

* There are two criteria that can be used to decompose our programs into modules:

>* **Decomposing by functionality**: instructions that perform similar tasks, along with the data they use, are put together. As an example, in the final project we would create a module for all the code moving game characters, another one for drawing things, another for checking collisions and so on. This is known as **structured programming** or **modular programming**: code is grouped into modules taking into account what it does.

>* **Decomposing by the entity they refer to**. In the final project we would put together all the instructions and data related to the players. Another module would contain all the code/data related to the packages. Another one could contain all the code/data related to the conveyor belts and so on. This is known as **object oriented** programming (OO). We can also notice that many of the code related to the players or the packages (how they behave, how to move them, etc.) will be similar. We will try to reuse as much of this code as possible, not repeating it.

* OO can be seen as an evolution of structured programming, but it is a different approach: we focus on entities/objects that are relevant. In OO we put together not only the code that is related to the same entity but also the variables related to it. 

* In OO:

>* each module is called a **class**
>* variables characterizing each object inside a class are called **attributes** or **fields**
>* code related to an object is split into **methods**
>* there will be many files where code is created and a *main* file where the code is executed.

* Advantages: OO provides a clear structure, easier to maintain because of the modular design and easier to be reused (this can also be obtained with Structured Programming, but OO makes it easier)

* Following an OO approach we analyze the problem in terms of:

1.	Entities that appear in our problem (classes/objects)
2.	Characteristics of those entities or information we need to store related to them (data attributes or fields)
3.	Things that those entities can do (methods)

* This is known as encapsulation (variables + methods operating on them): we put together data and code working over them. OO is also inheritance and polymorphism (we'll see them later in this topic)

# Creating a programming project using OOP

As said, complex programming projects are usually managed using an OOP approach. As with flow diagrams, it is usually good to use a graphic approach to identify which are the relevant entities in our program (classes), what data we need to store about them (attributes) and what they can do (methods). A widely used approach is the Unified Modeling Language (UML) that we will see in lab. We usually start with a class diagram:

1. Create a box for each of the types of entities you find in your program. For the final project we will have a box for Mario, packages, belts, truck...
2. For each of the entities think about the information you need to store about them. For Mario: image, level, side of the belt... Write them inside the box.
3. For each of the entities think about what can they do. For Mario, move up and down, take a package, rest... Write them inside the box.

Once you are done with that you can start writing your program. Each of the boxes will be a class, an inside it you will put the attributes (data) and the methods (what they can do).

# Classes

In OOP all the code related to an object will be created inside a special type of program named a *class*. All our classes will have the same structure: 
```
class NameOfTheClass:
    """Class documentation explaning what the class is doing and everything we
    must know about it """
    Code of the class
```

Naming conventions for class names: upper camel case (see above an example of a correct class name: `NameOfTheClass`)

It is recommended to put the class in its own file, do not use the same file for different classes nor include any other code in that file. The style rule says that the name of the file should be like the name of the class but starting by lower case.
 
It is very important to notice that inside a class no code can be alone. All the code inside a class must be grouped inside a special programming structure called *method*. If you have lines of code outside a method, your class is wrong. 

So, we can expand the class structure seen above to mark that anything in a class must be inside a method:

```
class NameOfTheClass:
    """Class documentation explaning what the class is doing and everything we
    must know about it """

    method1

    method2

    method3

    ...
```

## Methods

As we have seen, in object orientation there are two levels of decomposition:

>* Classes: here we put all the data and instructions related to an entity of our program.
>* Methods: inside a class the code is further decomposed into methods. They are small pieces of code that implement some things the object can do.

As an example into the Package class we may have a method to move it, another to check if it is handed by the player, another to check if it has fallen, etc.

A method is defined as follows:

```
def method_mame (self, param1, param2,...)
    """ This is the documentation of the method, it explains what the method
    is doing"""
    instruction1
    instruction2
    ...
```

A method has three parts:

>* The header `def method_name(self, param1, param2):` specifies the name of the method. It has also a pair of parenthesis that contain one or more names. All those names are names of variables. The first one is a special one, that we usually call `self` and is compulsory. The remaining ones are optional and contain the information needed to execute this code.
>* The documentation explains what the method is doing. It is also known as *docstring*. It is optional, but quite recommended. All our methods must have documentation this year.
>* The body are the instructions that are included in the method.

* Style rules (equal to variable names rules): names should be in small caps and if more than one word use underscores: `this_is_a_valid_method_name`. We can also use lower camel case but it less recommended: `thisIsALessRecommendedMethodName`

Taking that into account a class will look like this:

```
class NameOfTheClass:
    """ Class documentation """

    def method1 (self, param1, param2,...)
        """ This is the documentation of the method"""
        instruction1
        instruction2
        ...

    def method2 (self, param1, param2,...)
        """ This is the documentation of the method"""
        instruction1
        instruction2
        ...
```       

## The init method

The first thing we have to do when creating a new class it so think about which are the *attributes* of that class. In other words, which information I must store to represent the objects of the class. For example, for a package of the final project we could store: the image, the direction, the `x`, the `y`, if it is already at the truck...

As, in Python, everything inside a class must be written inside a method, the definition of the attributes must be performed also inside a method. A special method named the `__init__` method is used to declare the attributes.

It is one of the special or *magic* methods (magic methods usually start and end by `__`) 

It is used to declare which are the attributes of the objects of this class and to give values to them when an object is created. In other words, it is the code we use to specify which data we must store for an object.

It is similar to what in other languages is known as *constructor*.

```
def __init__(self, param1, param2...)
    """ Methods have at least 1 parameter, the 'self' that represents the whole
    object. In addition, init usually has a parameter for each attribute, 
    but there can be attributes that do not need a param """
    self.attribute1 = param1
    self.attribute2 = param2
    ...
    
```

Attributes are special variables declared inside the `init` by appending `self.` to their names. Name rules for attributes names: same as for variables (`self.this_is_a_valid_attribute_name`)

As a general rule this method must receive a parameter for each of the attributes of the object. 

>* But if the value of any attribute can be calculated using other attributes or it has always a fixed initial value or it is random, a parameter for that attribute is not needed.

Considering this, the general schema of a class will be:

```
class NameOfTheClass:
    """ Class documentation """
    
    def __init__(self, params_of_init):
        """ This method is used to declare the attributes"""
        # Attribute declaration
        self.attribute1 = value1
        self.attribute2 = value2
        ...

    def method1 (self, param1, param2,...)
        """ This is the documentation of the method"""
        instruction1
        instruction2
        ...

    def method2 (self, param1, param2,...)
        """ This is the documentation of the method"""
        instruction1
        instruction2
        ...
```      

## How to create the `init` method

After having drawn our class diagram to identify the attributes and methods of our objects, we can start our class by defining the attributes by creating the `__init__` method. To do it:

1. Start by creating the class and the header of the `__init__` method, until the `self`:

```
class MyClass:
    """ Class documentation """
    def __init__(self, 
        # Let this empty yet
```

2. Inside the `__init__`, create each attribute by putting the prefix `self.` to each of them:
```
class MyClass:
    """ Class documentation """
    def __init__(self, 
        self.attribute1 =
        self.attribute2 =
        self.attribute3 = 
```
3.  Now, think if to create the object you need to receive the value of the attribute, if so, put a parameter for that attribute in the header and assign the value of the parameter to the attribute. The name of the parameter is usually the same name than the attribute, but it is not compulsory:

```
class MyClass:
    """ Class documentation """
    # Here we assume just the values of attribute1 and attribute2 are needed
    def __init__(self, param1, param2)
        self.attribute1 = param1
        self.attribute2 = param2
        self.attribute3 = 
```
4.  If any attribute will start always with a given value or if its value will be random, just assign the initial value to the attribute (no parameter is needed for it):

```
class MyClass:
    """ Class documentation """
    # Here we assume just the values of attribute1 and attribute2 are needed
    def __init__(self, param1, param2)
        self.attribute1 = param1
        self.attribute2 = param2
        # This attribute will have always the same initial value
        self.attribute3 = initial_value
```

````python
# An example of a class with init method
class Date:
    """ This class describes the information I need to store about dates. It
    will be like a skeleton that I will use to create dates"""
    # In addition to self, init usually has parameters with values for each of the 
    # attributes. The name of the parameter is usually the same
    # name than the attribute, but it is not compulsory
    # It is a good idea to annotate each parameter with its type to help using the method
    def __init__(self, day: int, m: str, year: int):
        """ This method both declares the attributes of the class and
        receives the initial value for all them
        :param day: the day
        :param m: the month (called m to show that the name of parameter and attribute can be
                    different
        :param year: the year
        """
        # Attributes must be declared with self.name_of_attribute
        # In the init we usually just copy the value of each parameter into its
        # corresponding attribute
        self.day = day
        self.month = m
        self.year = year
````

# Objects
A class can be seen as a skeleton to create objects: it is like a generic definition of the variables (attributes or fields) and the instructions (methods) that an object will have. In other words, a class is a generic definition of all the objects of a given type.

A whole program will have two different kinds of modules: classes where fields and methods of the objects are defined and implemented, and a main program where objects belonging to those classes are used to build the algorithm. We define classes and work with objects.

Some important knowledge about objects:

* Classes just define the characteristics of the objects, to use a class we need to create an object.
* An object is a variable of a Class. In other words, an object is a variable whose type is a class.
* Values of the attributes of the new object must be specified when creating it. To create an object we use: `variable_name = ClassName(values_for_the_attributes)`. When doing it, we are really invoking the `init` method.
* To access or change the value of an attribute of an object we use *dot notation*: `object.attribute` to see the value, or `object.attribute = value` to change its value.
* What happens if we try to access a non-existing attribute? -> Error
* What happens if we try to assign a value to a non-existing attribute? -> The attribute is created (as in dictionaries). Do not create attributes on the fly. It is a bad programming practice as all the objects of a class must have the same attributes. It is very difficult to manage programs where each object of a class has different attributes. Totally **forbidden** this course.

````python
# This is the way to declare a variable belonging to the Date class
# Any variable belonging to a class is called an object: my_date is
# an object of the Date class
my_date = Date(day = 12, m = "october", year = 2024)
# To see the values to the attributes "dot notation" is used: object.attribute
# To recover the value of the day
print("The day is", my_date.day)
# Assigning a new value for the day
my_date.day = 22
print("The day is", my_date.day)
# For the month
my_date.month = "November"
print("The month is", my_date.month)
# For the year
my_date.year = 2021
print("The year is", my_date.year)
# This is another object of the Date class, notice we usually give values
# to parameters by their position instead of using their names
another_date = Date(22, "october", 2022)
````

## Types of variables inside a method

In the previous `init` methods we have seen two of the three types of variables that can appear inside a method:

1. **Attributes**: they store the data that characterizes an object. They are defined and used with `self.name`. They must be defined for the first time inside the `init` method but can be used by any other method (as we will see later).
2. **Parameters**: they are the variables defined in the header of a method and contain the information needed to execute the method. In the case of the `init` they contain the information required to give value to the attributes of the object, and usually have the same name as the attributes, but names could be different if we want. The parameters can only be used inside the method where they are defined. They do not exist in other methods.

It is important to notice that usually we need to put a parameter for each of the attributes except if:

1. The initial value of the attribute is always the same for all the objects.
2. The initial value of the attribute is rand

There is a third type of variable inside a method, the so called local variables: A **local variable** is a variable defined inside a method, that is used only for that method and does not exist in other methods. They are usually created to perform temporal calculations whose value is later stored on an attribute or returned (we will see that later). Notice the value of a local variable is lost when the method ends and cannot be used in any other part of the program. The way to distinguish attributes and local variables is by noticing that local variables do not start by `self.`

# Protecting object attributes

* Using an `init` method (constructor) we give the attributes an initial value. But nothing prevents us from giving to them values that make no sense or even values of a different type.
* All languages have mechanisms to avoid this, some of them just guide the programmer to do right things while other ones force right values by checking them and raising errors if they are not correct.
* In Python:

>* We can give information about the right values and types using the documentation of the class and parameter annotation.
>* We can force right values by using special methods called setters and properties. Other languages use private attributes (we will see them later) and methods to read their values and to change them (usually called `get` and `set` methods)

````python
# Creating objects with values for the attributes that make no sense
d = Date("hello", 4, "bye")
print("The date is", d.day, d.month, d.year)
# And I can also do it after creating them
d2 = Date(1, "november", 2022)
d2.day = "hello how are you?"
print("The date is", d2.day, d2.month, d2.year)
````

## Setters and properties

* Setters and properties are the way Python has to avoid the user of a class to give wrong values to the attributes (annotations are only suggestions)
* To use setters/properties in Python we need to do the following:

>1. In the `init` we create the attribute in the regular way `self.attribute = value`. We don't perform any checking of the value. We will also use the attribute in the regular way in any other method except for the property and the setter methods that we will create for each attribute.

>2. For each attribute for which we want to control the values, we create a new method with the same name of the attribute. This *magic* method is called *setter*. Its header will be `def attribute(self, value: type)` and it must be decorated with `@attribute.setter` in the line above the method. Inside this method I should check that the type of `value` is correct and also that its value is correct. If they are not right, exceptions will be risen. If the value is correct I will assign it to the attribute, but instead of doing `self.attribute = value` I must put two underscores before the name of the attribute: `self.__attribute = value`. If I don't do it I will have an error. This *magic* method will be automatically invoked any time I try to change the value of the attribute, either in the class or in the main program.

>3. If a setter was created like in the previous point, we are forced to create a *property*. A property is also a *magic* method. It must have too the name of the attribute and will receive only `self` as parameter: `def attribute(self) -> type`. It must be decorated with `@property` (exactly like this). In this special method all we can do is to return the value of the attribute adding two underscores before the name of the attribute: `return self.__attribute`. This will allow to read the value of the attribute but not to change it. The property must be placed before the setter.

* In `@property` and `@setter` we can also use `self._attribute` (only one underscore) instead of `self.__attribute` but if we do so, in addition to `attribute`, `_attribute` will be also visible outside and wrong values can be assigned to `attribute` by using `_attribute`, which is a non-desired side effect which makes the use of setters and properties almost useless (it will be penalized). 

* We'll see later that we can have a *property* without a *setter*, but setters can never be alone, if a setter is created a property must be defined first.

````python
# A Date class with setters and properties to avoid giving a wrong value to the fields
class Date:
    """A class to store a date information. We will use properties/setters to
    check the values provided are good. """

    def __init__(self, day: int, month: str, year: int):
        """ This method both declares the attributes of the class and
        receives the initial value for all them
        :param day: the day
        :param m: the month
        :param year: the year
        """
        # No checking is done here, everything is delegated to the setters
        # As I have a setter for year, this is totally equivalent to
        # invoke the year method: self.year(year)
        self.year = year
        # This also invokes the month setter
        self.month = month
        # As we need to have the month and the year to check the values of the
        # day, the day attribute must be declared after the year and month ones
        self.day = day
    
    # This @property keyword is called a 'decorator'
    @property
    # Notice we annotate the method with the type of the value it returns
    def year(self) -> int:
        """ This special method will return the value of the year
        :return : the year"""
        # Here I must return my attribute preceded by __
        # If I don't do I will have a maximum recursion error
        return self.__year
    
    # As this method's name is equal to the previous one, to avoid replacing 
    # the previous one, it needs to be decorated with @attribute.setter
    @year.setter
    def year(self, year: int):
        """ This method allows to change the value of the year
        :param year: the value of the year, if 0 is given an error is raisen"""
        # If the type is not correct we raise an exception
        if not isinstance(year, int):
            raise TypeError("The year  must be an int")
        # Here I need to use __year again 
        elif year != 0:
            self.__year = year
        else:
            # In any other case, I raise an exception
            raise ValueError("Year must be not equal to 0")            
    
    #We create properties and setters also for day and month
    @property
    def month(self) -> str:
        """ :return : the name of the month in small caps (as it will be
        set in small caps)"""
        return self.__month
    
    @month.setter
    def month(self, month: str):
        # Notice that inside a method we can use local variables.
        # Do not use self. with them or they will become attributes
        # and they will be visible outside the class. We only want this variable
        # to be used inside this method.
        days_in_month = {'january':31, 'february': 28, "march": 31, "april":30,
                         "may": 31, "june": 30, "july": 31, "august": 31,
                         "september": 30, "october": 31, "november": 30,
                         "december":31}
        if not isinstance(month, str):
            raise TypeError("The month must be a string")
        elif month.lower() in days_in_month:
            # We store the month name in lower case
            self.__month = month.lower()
        else:
            raise ValueError("Valid months are " + str(list(days_in_month)))
    
    @property
    def day(self) -> int:
        return self.__day
    
    @day.setter
    def day(self, day: int):
        # We cannot use a variable defined in another method, so we need to repeat
        # it here. Later we will see how to avoid this.
        days_in_month = {'january':31, 'february': 28, "march": 31, "april":30,
                         "may": 31, "june": 30, "july": 31, "august": 31,
                         "september": 30, "october": 31, "november": 30,
                         "december":31}
        # First checking the days of February
        # I change the days if leap year
        if self.year % 4 == 0 and (self.year % 100 != 0 or self.year % 400 == 0): 
            days_in_month['february'] = 29
        if not isinstance(day, int):
            raise TypeError("The day must be an integer")
        elif day > 0 and day <= days_in_month[self.month]:
            self.__day = day
        else:
            if self.month == 2:
                raise ValueError("February has only 28 days in " + str(self.year))  
            else:
                raise ValueError(self.month + " has only " + 
                                 str(days_in_month[self.month]) + " days")
````

## Raising exceptions

* In the previous example when a wrong value is received for an attribute we have risen an exception. Another possibility is setting it to a *safe* value.
* Raising an exception implies that an error will appear in my program and the program will finish.
* Whether to raise an exception of giving a safe value depends on the design: for some attributes it is better to provide safe values, for other ones it is more convenient to stop, as giving a safe value can lead to problems latter if the user is not aware of it. Raising errors is the recommended option and should be done this year.
* The most common raised exceptions when the value is wrong are:

>* `raise TypeError(error message)`: the type of the value is not the expected one
>* `raise ValueError(error message`: the type is correct, but the value is not

* The `error message` is optional but highly recommendable so the user knows why the program is failing.
* Exceptions make the program to stop, unless a `try-except` block is included (not covered in this course yet)

# Methods

* In addition to the `init` method, classes usually have other methods.
* Methods have always at least one parameter: the *self* parameter (we can call it whichever name we want, but `self` is the recommended one). This first and compulsory parameter of a method is a special one, because it represents the whole object. It contains a link to the attributes of the object so we can work with them inside the method.
* To invoke a method:

>* inside another method of the class I use `self.method_name(parameters without considering the self)`, in other words I omit the `self` parameter when invoking a method.
>* in the main program I use `object_name.method_name(parameters without considering the self)`, we also omit the `self` parameter.

* Methods must either change the value of one or more attributes or do some calculation with the values of the attributes and return it, or both behaviors. As we have seen in the properties and setters the branching instruction `return` is used to return the value. This instruction also makes the method to finish.

````python
# Adding a method to our Date class 
class Date:
    """A class to store a date information. We will use properties/setters to
    check the values provided are good. It has also a method to return the number
    of the month"""

    def __init__(self, day: int, month: str, year: int):
        """ This method both declares the attributes of the class and
        receives the initial value for all them
        :param day: the day
        :param m: the month
        :param year: the year
        """
        # No checking is done here, everything is delegated to the setters
        # As I have a setter for year, this is totally equivalent to
        # invoke the year method: self.year(year)
        self.year = year
        # This also invokes the month setter
        self.month = month
        # As we need to have the month and the year to check the values of the
        # day, the day attribute must be declared after the year and month ones
        self.day = day
    
    # This @property keyword is called a 'decorator'
    @property
    # Notice we annotate the method with the type of the value it returns
    def year(self) -> int:
        """ This special method will return the value of the year
        :return : the year"""
        # Here I must return my attribute preceded by __
        # If I don't do I will have a maximum recursion error
        return self.__year
    
    # As this method's name is equal to the previous one, to avoid replacing 
    # the previous one, it needs to be decorated with @attribute.setter
    @year.setter
    def year(self, year: int):
        """ This method allows to change the value of the year
        :param year: the value of the year, if 0 is given an error is raisen"""
        # If the type is not correct we raise an exception
        if not isinstance(year, int):
            raise TypeError("The year  must be an int")
        # Here I need to use __year again 
        elif year != 0:
            self.__year = year
        else:
            # In any other case, I raise an exception
            raise ValueError("Year must be not equal to 0")            
    
    #We create properties and setters also for day and month
    @property
    def month(self) -> str:
        """ :return : the name of the month in small caps (as it will be
        set in small caps)"""
        return self.__month
    
    @month.setter
    def month(self, month: str):
        # Notice that inside a method we can use local variables.
        # Do not use self. with them or they will become attributes
        # and they will be visible outside the class. We only want this variable
        # to be used inside this method.
        days_in_month = {'january':31, 'february': 28, "march": 31, "april":30,
                         "may": 31, "june": 30, "july": 31, "august": 31,
                         "september": 30, "october": 31, "november": 30,
                         "december":31}
        if not isinstance(month, str):
            raise TypeError("The month must be a string")
        elif month.lower() in days_in_month:
            # We store the month name in lower case
            self.__month = month.lower()
        else:
            raise ValueError("Valid months are " + str(list(days_in_month)))
    
    @property
    def day(self) -> int:
        return self.__day
    
    @day.setter
    def day(self, day: int):
        # We cannot use a variable defined in another method, so we need to repeat
        # it here. Later we will see how to avoid this.
        days_in_month = {'january':31, 'february': 28, "march": 31, "april":30,
                         "may": 31, "june": 30, "july": 31, "august": 31,
                         "september": 30, "october": 31, "november": 30,
                         "december":31}
        # First checking the days of February
        # I change the days if leap year
        if self.year % 4 == 0 and (self.year % 100 != 0 or self.year % 400 == 0): 
            days_in_month['february'] = 29
        if not isinstance(day, int):
            raise TypeError("The day must be an integer")
        elif day > 0 and day <= days_in_month[self.month]:
            self.__day = day
        else:
            if self.month == 2:
                raise ValueError("February has only 28 days in " + str(self.year))  
            else:
                raise ValueError(self.month + " has only " + 
                                 str(days_in_month[self.month]) + " days")
                
    def month_number(self) -> int:
        """ A method that returns the number of the month.
        :return: a value between 1 and 12"""
        # A local variable to store the months in order
        months = ('january', 'february', "march", "april","may", "june", "july", "august",
                  "september", "october", "november", "december")
        for pos in range(len(months)):
            if self.month == months[pos]:
                number = pos + 1   
        return number
````

# More magic methods
* In addition to `__init__` there are many other magic methods, we usually implement some of them in all our classes.
* The name of all of them starts and ends by a double underscore: `__name__`
* Some of the most relevant ones are:

>* The `__str__(self)` method allows to specify what should be shown if an object is printed
>* The `__repr__(self)` method allows to specify what should be shown if the name of an object is typed in the interpreter.
>* The `__eq__(self, another_object)` method allows to compare two objects using `==`. If it is not implemented Python compares the memory addresses. We also usually implement `__ne__(self, other)` for `!=`, `__lt__(self, other)` for `<`, `__gt__(self, other)` for `>`, `__le__(self, other)`, for `<=`, and `__ge__(self, other)` for `>=`.

Notice that usually is enough to implement `__eq__`, `__lt__` and `__le__`, as `__ne__`,   `__gt__` and `__ge__` are working just in the opposite way.

````python
# Adding more magic methods to our Date class 
class Date:
    """A class to store a date information. We will use properties/setters to
    check the values provided are good. It has also a method to return the number
    of the month"""

    def __init__(self, day: int, month: str, year: int):
        """ This method both declares the attributes of the class and
        receives the initial value for all them
        :param day: the day
        :param m: the month
        :param year: the year
        """
        # No checking is done here, everything is delegated to the setters
        # As I have a setter for year, this is totally equivalent to
        # invoke the year method: self.year(year)
        self.year = year
        # This also invokes the month setter
        self.month = month
        # As we need to have the month and the year to check the values of the
        # day, the day attribute must be declared after the year and month ones
        self.day = day
    
    # This @property keyword is called a 'decorator'
    @property
    # Notice we annotate the method with the type of the value it returns
    def year(self) -> int:
        """ This special method will return the value of the year
        :return : the year"""
        # Here I must return my attribute preceded by __
        # If I don't do I will have a maximum recursion error
        return self.__year
    
    # As this method's name is equal to the previous one, to avoid replacing 
    # the previous one, it needs to be decorated with @attribute.setter
    @year.setter
    def year(self, year: int):
        """ This method allows to change the value of the year
        :param year: the value of the year, if 0 is given an error is raisen"""
        # If the type is not correct we raise an exception
        if not isinstance(year, int):
            raise TypeError("The year  must be an int")
        # Here I need to use __year again 
        elif year != 0:
            self.__year = year
        else:
            # In any other case, I raise an exception
            raise ValueError("Year must be not equal to 0")            
    
    #We create properties and setters also for day and month
    @property
    def month(self) -> str:
        """ :return : the name of the month in small caps (as it will be
        set in small caps)"""
        return self.__month
    
    @month.setter
    def month(self, month: str):
        # Notice that inside a method we can use local variables.
        # Do not use self. with them or they will become attributes
        # and they will be visible outside the class. We only want this variable
        # to be used inside this method.
        days_in_month = {'january':31, 'february': 28, "march": 31, "april":30,
                         "may": 31, "june": 30, "july": 31, "august": 31,
                         "september": 30, "october": 31, "november": 30,
                         "december":31}
        if not isinstance(month, str):
            raise TypeError("The month must be a string")
        elif month.lower() in days_in_month:
            # We store the month name in lower case
            self.__month = month.lower()
        else:
            raise ValueError("Valid months are " + str(list(days_in_month)))
    
    @property
    def day(self) -> int:
        return self.__day
    
    @day.setter
    def day(self, day: int):
        # We cannot use a variable defined in another method, so we need to repeat
        # it here. Later we will see how to avoid this.
        days_in_month = {'january':31, 'february': 28, "march": 31, "april":30,
                         "may": 31, "june": 30, "july": 31, "august": 31,
                         "september": 30, "october": 31, "november": 30,
                         "december":31}
        # First checking the days of February
        # I change the days if leap year
        if self.year % 4 == 0 and (self.year % 100 != 0 or self.year % 400 == 0): 
            days_in_month['february'] = 29
        if not isinstance(day, int):
            raise TypeError("The day must be an integer")
        elif day > 0 and day <= days_in_month[self.month]:
            self.__day = day
        else:
            if self.month == 2:
                raise ValueError("February has only 28 days in " + str(self.year))  
            else:
                raise ValueError(self.month + " has only " + 
                                 str(days_in_month[self.month]) + " days")
                
    def month_number(self) -> int:
        """ A method that returns the number of the month.
        :return: a value between 1 and 12"""
        # A local variable to store the months in order
        months = ('january', 'february', "march", "april","may", "june", "july", "august",
                  "september", "october", "november", "december")
        for pos in range(len(months)):
            if self.month == months[pos]:
                number = pos + 1   
        return number
        
    def __str__(self) -> str:
        """This method is invoked if I print the object. It must
        be called like this, no changes in name, no extra parameters.
        It must return the string I want to be shown when printing"""
        suffix = "th"
        if self.day == 1 or self.day == 21 or self.day == 31:
            suffix = "st"
        elif self.day == 2 or self.day == 22:
            suffix = "nd"
        elif self.day == 3 or self.day == 23:
            suffix = "rd"
        # First letter of the month in capital 
        month = self.month[0].upper() +  self.month[1:]
        return month + " " + str(self.day) + suffix +", " + str(self.year)
    
    def __repr__(self) -> str:
        """This method is invoked if I write the name of the object in
        the interpreter. It must
        be called like this, no changes in name, no extra parameters.
        It must return the string I want to be shown"""
        # I usually just invoke the str
        return self.__str__()
    
    def __eq__(self, other) -> bool:
        """ This is the method Python will invoke if I try to compare
        two Date objects. In addition to self it must receive the other
        object. It must return True or False"""
        # If the type of the other object is not Date the convention says
        # to return NotImplemented. 
        if isinstance(other, Date):
        # Two dates are equal if the values of all their attributes are
        # equal
            return (self.day == other.day and self.month == other.month 
                        and self.year == other.year)
        else:
            return NotImplemented
    
    def __lt__(self, other) -> bool:
        """ This is the method Python will invoke if < is used with a Date object """
        if isinstance(other, Date):
            # We check if the year is <
            return (self.year < other.year or
                    # Or if the year is equal and the month < 
                    # To compare months we use their number, notice the way the month_number() method
                    # is invoked: self.method()
                   (self.year == other.year and self.month_number() < other.month_number()) or
                    # Or if the day is <
                   (self.year == other.year and self.month_number() == other.month_number() and self.day < other.day))
        else:
            return NotImplemented

    def __le__(self, other) -> bool:
        """ This is the method Python will invoke if the <= is used with a Date object """
        if isinstance(other, Date):
            # Reusing the previous methods
            return self < other or self == other
        else:
            return NotImplemented
````

# Composition

* In OOP there are two ways to create a class using another class: composition and inheritance.

* In composition one or more of the attributes of a class is an object of another class.

````python
class Student:
    """ A class to represent a student. It uses composition"""
    def __init__(self, name: str, surname: str, birthday: Date):
        """
        :param name: the name of the student
        :param surname: the surname of the student
        :param birthday: the birthday
        """
        self.name = name
        self.surname = surname
        # birthday is an object of Date class
        self.birthday = birthday
        
    @property
    def name(self) -> str:
        return self.__name
    @property
    def surname(self) -> str:
        return self.__surname
    @property
    def birthday(self) -> Date:
        return self.__birthday

    # When an attribute is of type str and any value is valid, we usually do not
    # provide properties/setters, but we will do it here to practice 
    @name.setter
    def name(self, name:str):
        # Just the type is checked
        if not isinstance(name, str):
            raise TypeError("The name must be a str")
        else:
            self.__name = name
    
    @surname.setter
    def surname(self, surname:str):
        if not isinstance(surname, str):
            raise TypeError("The surname must be a str")
        else:
            self.__surname = surname            
    
    @birthday.setter
    def birthday(self, bd):
        # We just check if the type is correct
        # Checking if the values are correct is delegated to the Date class
        if type(bd) != Date:
            raise TypeError("The birthday must be a date")
        else:
            self.__birthday = bd
    
    def __str__(self) -> str:
        return self.name + " " + self.surname + " born on " + str(self.birthday)

    def __repr__(self) -> str:
        return self.__str__()

    def __eq__(self, other) -> bool:
        if isinstance(other, Student):
            return self.name == other.name and self.surname == other.surname and self.birthday == other.birthday
        else:
            return NotImplemented
    
    def __lt__(self, other) -> bool:
        if isinstance(other, Student): 
            # A Student is < than another if it is younger, so its birthday
            # date is >
            return self.birthday > other.birthday
        else:
            return NotImplemented

    def __le__(self, other) -> bool:
        if isinstance(other, Student):  
            return self.birthday >= other.birthday
        else:
            return NotImplemented            
````

````python
# First way to create a composed object
# We use an auxiliary variable
d = Date(1, "december", 1984)
st = Student("Pepe", "Perez", d)
print(st)
# The name of the student
print("The name of the student is", st.name)
# The day the student was born
print("His/her born day is", st.birthday.day)
# Second way to create a composed object, on the fly (recommended)
st2 = Student("Luisa", "Gomez", Date(1, "december", 2002))
print(st2)
print("Are", st, "and", st2, "equal?", st == st2)
print("Is", st, ">", st2, "?", st > st2)
````

````python
class Board:
    def __init__(self, width: int, height: int):
        """ Creates a board for a game with an Enemy placed at the middle of it.
        :param width: the width of the board
        :param height: the height of the board
        """
        self.width = width
        self.height = height
        # In this case as we know the initial values of the Package attributes,
        # we don't need to receive a parameter for it
        self.package = Package(0, "image_package_level_0.png")

    @property
    def width(self) -> int:
        return self.__width
    @property
    def height(self) -> int:
        return self.__height
    @property
    def package(self) -> Package:
        return self.__package

    @width.setter
    def width(self, width: int):
        if not isinstance(width, int):
            raise TypeException ("The width must be an int")
        elif width <= 0:
            raise ValueException("No negative or zero width allowed")
        else:
            self.__width = width

    @height.setter
    def height(self, height: int):
        if not isinstance(height, int):
            raise TypeException ("The height must be an int")
        elif height <= 0:
            raise ValueException("No negative or zero height allowed")
        else:
            self.__height = height

    @package.setter
    def package(self, package: Package):
        if not isinstance(package, Package):
            raise TypeError("The package must belong to the Package class")
        else:
            self.__package = package

    def __str__(self) -> str:
        """ Shows the board with a P in the place where the package is"""
        board_list = []
        for row in range(self.height):
            board_list.append([])
            for col in range(self.width):
                board_list[-1].append(" ")
        # Indexes start in 0 so we substract 1, also the y is the row and the x the column
        # Here we consider the (0,0) the bottom right
        # corner. x increases to the left and y to the top. Notice pyxel
        # considers the (0, 0) as the top right corner.
        board_list[self.height - 1 - self.package.y][self.width - 1 - self.package.x] = "P"
        res = ""
        for sub_list in board_list:
            res += str(sub_list) + "\n"
        return res        
````

````python
# Another example of composition, an attribute is a list of objects
class Classroom:
    def __init__(self, students: list[Student]):
        self.students = students

    @property
    def students(self) -> list[Student]:
        return self.__students

    @students.setter
    def students(self, st):
        if type(st) != list:
            raise TypeError("The students must be a list")
        else:
            for element in st:
                if type(element) != Student:
                    raise TypeError("All the elements of the list must be of type Student")
        # If this line is reached, everything is correct
        self.__students = st

    # Wrong attempt to perform the __str__
    # def __str__(self) -> str:
    #     for student in self.students:
    #         # This will only return the name of the first student, as return
    #         # means return and finish
    #         return str(student)
    
    def __str__(self) -> str:
        """ Returns all the students, each one on its own line """
        stri = ""
        for z in self.students:
            stri += (str(z) + "\n")
            # We could also do stri += z.__str__() + "\n"
        return stri
    
    def __eq__(self, other) -> bool:
        """ Two classroms are equal if their list of students is the same"""
        if not isinstance(other, Classroom):
            return NotImplemented
        else:
            return self.students == other.students

    def __lt__(self, other) -> bool:
        """ A classroom is < than other if its number of students is <"""
        if not isinstance(other, Classroom):
            return NotImplemented
        else:
            return len(self.students) < len(other.students)

    def __le__(self, other) -> bool:
        if not isinstance(other, Classroom):
            return NotImplemented
        else:
            return len(self.students) <= len(other.students)
````

# Encapsulation: Information hiding

As we have seen, in OO we put together all the information related to objects of one type along with the methods needed to work with them. This is called encapsulation and it is one of the three characteristics of OO (the other two ones are *inheritance* and *polymorphism*).

One advantage of encapsulation is that we can hide the way classes are implemented, making not visible outside the class some attributes and methods, or making some attributes to be read only so their values cannot be changed

## Read-only attributes

It is quite common to have some attributes whose value we want to read but we don't want to allow any program to change. 

In Python we get this behavior by a having a property without setter. This allows us to have new attributes whose values are calculated using other attributes.

````python
# A Date class with read-only attribute
class Date:
    """A class to store a date information. We will use properties/setters to
    check the values provided are good. It has also a method to return the number
    of the month"""

    def __init__(self, day: int, month: str, year: int):
        """ This method both declares the attributes of the class and
        receives the initial value for all them
        :param day: the day
        :param m: the month
        :param year: the year
        """
        # No checking is done here, everything is delegated to the setters
        # As I have a setter for year, this is totally equivalent to
        # invoke the year method: self.year(year)
        self.year = year
        # This also invokes the month setter
        self.month = month
        # As we need to have the month and the year to check the values of the
        # day, the day attribute must be declared after the year and month ones
        self.day = day
    
    # This @property keyword is called a 'decorator'
    @property
    # Notice we annotate the method with the type of the value it returns
    def year(self) -> int:
        """ This special method will return the value of the year
        :return : the year"""
        # Here I must return my attribute preceded by __
        # If I don't do I will have a maximum recursion error
        return self.__year
    
    # As this method's name is equal to the previous one, to avoid replacing 
    # the previous one, it needs to be decorated with @attribute.setter
    @year.setter
    def year(self, year: int):
        """ This method allows to change the value of the year
        :param year: the value of the year, if 0 is given an error is raisen"""
        # If the type is not correct we raise an exception
        if not isinstance(year, int):
            raise TypeError("The year  must be an int")
        # Here I need to use __year again 
        elif year != 0:
            self.__year = year
        else:
            # In any other case, I raise an exception
            raise ValueError("Year must be not equal to 0")            

    #This is a read-only attribute that we can read inside the class or outside it
    # Notice it has a property but not a setter, so its value cannot be set
    @property
    def leap_year(self) -> bool:
        """ The value of leap_year is calculated using other attributes"""
        return self.year % 4 == 0 and (self.year % 100 != 0 or self.year % 400 == 0) 
    
    #We create properties and setters also for day and month
    @property
    def month(self) -> str:
        """ :return : the name of the month in small caps (as it will be
        set in small caps)"""
        return self.__month
    
    @month.setter
    def month(self, month: str):
        # Notice that inside a method we can use local variables.
        # Do not use self. with them or they will become attributes
        # and they will be visible outside the class. We only want this variable
        # to be used inside this method.
        days_in_month = {'january':31, 'february': 28, "march": 31, "april":30,
                         "may": 31, "june": 30, "july": 31, "august": 31,
                         "september": 30, "october": 31, "november": 30,
                         "december":31}
        if not isinstance(month, str):
            raise TypeError("The month must be a string")
        elif month.lower() in days_in_month:
            # We store the month name in lower case
            self.__month = month.lower()
        else:
            raise ValueError("Valid months are " + str(list(days_in_month)))
    
    @property
    def day(self) -> int:
        return self.__day
    
    @day.setter
    def day(self, day: int):
        # We cannot use a variable defined in another method, so we need to repeat
        # it here. Later we will see how to avoid this.
        days_in_month = {'january':31, 'february': 28, "march": 31, "april":30,
                         "may": 31, "june": 30, "july": 31, "august": 31,
                         "september": 30, "october": 31, "november": 30,
                         "december":31}
        # First checking the days of February
        # I change the days if leap year, we use the read only attribute
        if self.leap_year: 
            days_in_month['february'] = 29
        if not isinstance(day, int):
            raise TypeError("The day must be an integer")
        elif day > 0 and day <= days_in_month[self.month]:
            self.__day = day
        else:
            if self.month == 2:
                raise ValueError("February has only 28 days in " + str(self.year))  
            else:
                raise ValueError(self.month + " has only " + 
                                 str(days_in_month[self.month]) + " days")
                
    def month_number(self) -> int:
        """ A method that returns the number of the month.
        :return: a value between 1 and 12"""
        # A local variable to store the months in order
        months = ('january', 'february', "march", "april","may", "june", "july", "august",
                  "september", "october", "november", "december")
        for pos in range(len(months)):
            if self.month == months[pos]:
                number = pos + 1   
        return number
        
    def __str__(self) -> str:
        """This method is invoked if I print the object. It must
        be called like this, no changes in name, no extra parameters.
        It must return the string I want to be shown when printing"""
        suffix = "th"
        if self.day == 1 or self.day == 21 or self.day == 31:
            suffix = "st"
        elif self.day == 2 or self.day == 22:
            suffix = "nd"
        elif self.day == 3 or self.day == 23:
            suffix = "rd"
        # First letter of the month in capital 
        month = self.month[0].upper() +  self.month[1:]
         # We also change the str to show if the year is leap or not
        leap = " (non-leap)"
        if self.leap_year:
            leap = " (leap)"
        return month + " " + str(self.day) + suffix +", " + str(self.year) + leap
    
    def __repr__(self) -> str:
        """This method is invoked if I write the name of the object in
        the interpreter. It must
        be called like this, no changes in name, no extra parameters.
        It must return the string I want to be shown"""
        # I usually just invoke the str
        return self.__str__()
    
    def __eq__(self, other) -> bool:
        """ This is the method Python will invoke if I try to compare
        two Date objects. In addition to self it must receive the other
        object. It must return True or False"""
        # If the type of the other object is not Date the convention says
        # to return NotImplemented. 
        if isinstance(other, Date):
        # Two dates are equal if the values of all their attributes are
        # equal
            return (self.day == other.day and self.month == other.month 
                        and self.year == other.year)
        else:
            return NotImplemented
    
    def __lt__(self, other) -> bool:
        """ This is the method Python will invoke if < is used with a Date object """
        if isinstance(other, Date):
            # We check if the year is <
            return (self.year < other.year or
                    # Or if the year is equal and the month < 
                    # To compare months we use their number, notice the way the month_number() method
                    # is invoked: self.method()
                   (self.year == other.year and self.month_number() < other.month_number()) or
                    # Or if the day is <
                   (self.year == other.year and self.month_number() == other.month_number() and self.day < other.day))
        else:
            return NotImplemented

    def __le__(self, other) -> bool:
        """ This is the method Python will invoke if the <= is used with a Date object """
        if isinstance(other, Date):
            # Reusing the previous methods
            return self < other or self == other
        else:
            return NotImplemented
````
